#Experiment-1.1: Read the employee dataset and display the average salary of each department using a bar chart.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/drive/MyDrive/CNLP-Lab/Exp-1/employee_information_100.csv')

avg_salary = df.groupby('Department')['Salary'].mean()

plt.bar(avg_salary.index, avg_salary.values)
plt.xlabel('Department')
plt.ylabel('Average Salary')
plt.title('Average Salary by Department')
plt.show()
```

#Experiment 1.2: Display the number of employees working in each department.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/drive/MyDrive/CNLP-Lab/Exp-1/employee_information_100.csv')

dept_count = df['Department'].value_counts()

print(dept_count)

plt.bar(dept_count.index, dept_count.values)
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.title('Number of Employees by Department')
plt.show()
```

#Experiment 1.3:Display the percentage of male and female employees using a pie chart


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/drive/MyDrive/CNLP-Lab/Exp-1/employee_information_100.csv')

gender_count = df['Gender'].value_counts()

print(gender_count)

plt.pie(
    gender_count.values,
    labels=gender_count.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Percentage of Male and Female Employees')
plt.show()
```

#Experiment 1.4: Draw a histogram showing salary distribution.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/drive/MyDrive/CNLP-Lab/Exp-1/employee_information_100.csv')

plt.hist(df['Salary'], bins=10, edgecolor='black')

plt.xlabel('Salary')
plt.ylabel('Number of Employees')
plt.title('Salary Distribution')
plt.show()
```

#Experiment 1.8: Display employees whose salary is greater than the overall average salary.
