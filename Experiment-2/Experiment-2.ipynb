#Experiment 2.1:  Write a Python program to perform basic text preprocessing on a given data. The program should:
#1.Convert text to lowercase. (Also display how many uppercase?)
#2.Remove punctuation.        (Also display how many different types of punctuations?)
#3.Remove numbers.            (Also display how many numbers?)
#4.Remove extra whitespace. (How much extra whitespace is present?)
#5.Display the cleaned text.


```python
import string
import re

# Read the text file
with open("/content/drive/MyDrive/CNLP-Lab/Exp-2/2.1_text_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Display original text
print("Original Text:")
print(text)

# ---------------------------------------------------------
# 1. Count uppercase letters and convert text to lowercase
# ---------------------------------------------------------

uppercase_count = sum(1 for char in text if char.isupper())

text = text.lower()

print("\nNumber of uppercase letters:", uppercase_count)


# ---------------------------------------------------------
# 2. Count and remove punctuation
# ---------------------------------------------------------

# Find all punctuation characters
punctuations = [char for char in text if char in string.punctuation]

# Find different types of punctuation
different_punctuations = set(punctuations)

print("Total punctuation marks:", len(punctuations))
print("Different types of punctuation:", len(different_punctuations))
print("Punctuation types:", different_punctuations)

# Remove punctuation
text = text.translate(str.maketrans("", "", string.punctuation))


# ---------------------------------------------------------
# 3. Count and remove numbers
# ---------------------------------------------------------

# Count individual digits
digit_count = sum(1 for char in text if char.isdigit())

# Count complete numbers
numbers = re.findall(r'\d+', text)
number_count = len(numbers)

print("Total digits:", digit_count)
print("Number of numbers:", number_count)

# Remove numbers
text = re.sub(r'\d+', '', text)


# ---------------------------------------------------------
# 4. Count and remove extra whitespace
# ---------------------------------------------------------

# Count whitespace before cleaning
original_whitespace = len(re.findall(r'\s', text))

# Replace multiple whitespace characters with one space
cleaned_whitespace = re.sub(r'\s+', ' ', text).strip()

# Calculate extra whitespace removed
extra_whitespace = original_whitespace - len(re.findall(r'\s', cleaned_whitespace))

print("Extra whitespace:", extra_whitespace)

text = cleaned_whitespace


# ---------------------------------------------------------
# 5. Display cleaned text
# ---------------------------------------------------------

print("\nCleaned Text:")
print(text)
```

#Experiment 2.2: Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.
#           (Three separate code: using NLTK, using spaCy  and without using NLTK/spacy)


#I.Using NLTK

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')

# Read the text file
with open("/content/drive/MyDrive/CNLP-Lab/Exp-2/2.2_tokenization_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Sentence Tokenization
sentences = sent_tokenize(text)

# Word Tokenization
words = word_tokenize(text)

# Display sentences
print("----- SENTENCE TOKENS -----")

for i, sentence in enumerate(sentences, start=1):
    print(i, ":", sentence)

# Display words
print("\n----- WORD TOKENS -----")

for i, word in enumerate(words, start=1):
    print(i, ":", word)
```

#II. Using spaCY

```python
import spacy

# Load English language model
nlp = spacy.load("en_core_web_sm")

# Read the text file
with open("/content/drive/MyDrive/CNLP-Lab/Exp-2/2.2_tokenization_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Process the text
doc = nlp(text)

# Sentence Tokenization
print("----- SENTENCE TOKENS -----")

for i, sentence in enumerate(doc.sents, start=1):
    print(i, ":", sentence.text)

# Word Tokenization
print("\n----- WORD TOKENS -----")

for i, token in enumerate(doc, start=1):
    print(i, ":", token.text)
```

#III. Without NLTK or spaCy

```python
import re

# Read the text file
with open("/content/drive/MyDrive/CNLP-Lab/Exp-2/2.2_tokenization_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

# -------------------------------------------------
# Sentence Tokenization
# -------------------------------------------------

sentences = re.split(r'(?<=[.!?])\s+', text.strip())

print("----- SENTENCE TOKENS -----")

for i, sentence in enumerate(sentences, start=1):
    print(i, ":", sentence)


# -------------------------------------------------
# Word Tokenization
# -------------------------------------------------

words = re.findall(r'\b\w+\b|[^\w\s]', text)

print("\n----- WORD TOKENS -----")

for i, word in enumerate(words, start=1):
    print(i, ":", word)
```

#Experiment 2.3: Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).


```python
import nltk
from nltk.tokenize import word_tokenize

# Download required tokenizer data
nltk.download('punkt')
nltk.download('punkt_tab')

# File path
file_path = "/content/drive/MyDrive/CNLP-Lab/Exp-2/2.3_clean_data.txt"

# Read text from the file
with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

# Define stop words
stop_words = {
    "is", "the", "a", "an", "in", "of", "and", "to",
    "this", "it", "on", "for", "with", "as", "at", "by"
}

# Tokenize the text
tokens = word_tokenize(text)

# Find stop words
found_stop_words = [
    word for word in tokens
    if word.lower() in stop_words
]

# Remove stop words
filtered_tokens = [
    word for word in tokens
    if word.lower() not in stop_words
]

# Display results
print("Original Tokens:")
print(tokens)

print("\nStop Words:")
print(found_stop_words)

print("\nFiltered Tokens:")
print(filtered_tokens)
```
